```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Charger le dataset
df_brut = pd.read_csv("diabetes_brut.csv")

# Afficher les premières lignes
df_brut.head()
```

```python

# Les 5 premières lignes
display(df_brut.head())

# Les 3 dernières lignes
display(df_brut.tail(3))

# Dimensions
print("Dimensions :", df_brut.shape)

# Noms des colonnes
print("Colonnes :", df_brut.columns.tolist())

# Types des colonnes
print("\nTypes des colonnes :")
print(df_brut.dtypes)
```

```python

df_brut.info()
```

```python

df_brut.describe().round(1)
```

```python

# Nombre de valeurs manquantes par colonne
print("Valeurs manquantes :")
print(df_brut.isnull().sum())

# Nombre de doublons sur toutes les colonnes
print("\nDoublons complets :")
print(df_brut.duplicated().sum())

# Nombre de doublons selon la colonne ID
print("\nDoublons selon ID :")
print(df_brut.duplicated(subset=["id"]).sum())
```

```python

colonnes = ["id", "age", "sexe", "imc", "progression"]

doublons = df_brut.duplicated(keep=False)

display(df_brut.loc[doublons, colonnes])
```

```python

# Créer une copie sans doublons
df = df_brut.drop_duplicates().copy()

print("Dimensions avant :", df_brut.shape)
print("Dimensions après :", df.shape)
```

```python

df["pa"] = pd.to_numeric(df["pa"], errors="coerce")

print("Valeurs non convertibles :")
print(df["pa"].isna().sum())
```

```python

# Détecter les âges supérieurs à 120
masque = df["age"] > 120

# Afficher les patients concernés
display(df.loc[masque, ["id", "age", "imc"]])

# Remplacer les âges incohérents par NaN
df.loc[masque, "age"] = np.nan

print("Nombre d'âges incohérents :", masque.sum())
```

```python

print("Avant suppression :", len(df))

df = df.dropna(subset=["age", "imc"])

print("Après suppression :", len(df))
```

```python

# Calculer la médiane
mediane_pa = df["pa"].median()

print("Médiane de PA :", mediane_pa)

# Remplacer les valeurs manquantes
df["pa"] = df["pa"].fillna(mediane_pa)

# Vérifier
print("Valeurs manquantes de PA :", df["pa"].isna().sum())
print("Valeurs manquantes de GLU :", df["glu"].isna().sum())
```

```python

df["sexe"] = df["sexe"].astype("category")
df["pa"] = df["pa"].astype(float)

print(df.dtypes)
```

```python

bins_age = [18, 35, 50, 65, np.inf]

labels_age = [
    "19-35",
    "36-50",
    "51-65",
    "66 ans et plus"
]

df["classe_age"] = pd.cut(
    df["age"],
    bins=bins_age,
    labels=labels_age
)
```

```python

bins_imc = [-np.inf, 25, 30, np.inf]

labels_imc = [
    "moins de 25",
    "de 25 à 30",
    "30 ou plus"
]

df["categorie_imc"] = pd.cut(
    df["imc"],
    bins=bins_imc,
    labels=labels_imc,
    right=False
)
```

```python

print("Répartition par classe d'âge :")
print(df["classe_age"].value_counts(sort=False))

print("\nRépartition par catégorie d'IMC :")
print(df["categorie_imc"].value_counts(sort=False))
```

```python

print("Dimensions finales :", df.shape)

display(df.head())

print("\nValeurs manquantes :")
print(df.isnull().sum())

print("\nTypes des colonnes :")
print(df.dtypes)
```

```python

imc_moyen = df.groupby(
    "classe_age",
    observed=True
)["imc"].mean()

print(imc_moyen)
```

```python

plt.figure(figsize=(7, 4))

plt.plot(
    imc_moyen.index,
    imc_moyen.values,
    marker="o",
    linestyle="-"
)

plt.title("IMC moyen par classe d'âge")
plt.xlabel("Classe d'âge")
plt.ylabel("IMC moyen (kg/m²)")

plt.grid(True, linestyle="--")

plt.tight_layout()

plt.savefig("imc_moyen_par_age.png", dpi=300)

plt.show()
```

```python

tableau = pd.crosstab(
    df["classe_age"],
    df["sexe"]
)

print(tableau)
```

```python

prop = pd.crosstab(
    df["sexe"],
    df["categorie_imc"],
    normalize="index"
) * 100

print(prop)
```

```python

plt.figure(figsize=(8, 5))

bas = np.zeros(len(prop))

for cat in prop.columns:
    plt.bar(
        prop.index.astype(str),
        prop[cat].values,
        bottom=bas,
        label=cat
    )
    bas = bas + prop[cat].values

plt.title("Répartition des catégories d'IMC selon le sexe")
plt.xlabel("Sexe")
plt.ylabel("Pourcentage (%)")

plt.legend(title="Catégorie d'IMC")
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```
